```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# Circuito cuántico simple donde estimamos la amplitud del estado |1>
qc = QuantumCircuit(1)
qc.ry(1.0, 0)   # rotación que genera una cierta amplitud en |1>
qc.measure_all()

# Usamos el simulador moderno
sim = AerSimulator()

# Ejecutamos el circuito
result = sim.run(qc, shots=2000).result()

# Obtenemos el conteo de estados
counts = result.get_counts()
counts
```
